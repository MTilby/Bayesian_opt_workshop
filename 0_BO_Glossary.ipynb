# BO Glossary

A short glossary for the terms used across workbooks 1-5.
This can be used as a reference for the other workbooks.

## Core BO terms

| Term | Meaning |
| --- | --- |
| Bayesian Optimisation (BO) | A probabilistic strategy for choosing the next experiment when experiments are expensive and data are limited. |
| Objective | The quantity you want to improve, such as yield or productivity. |
| Maximise / minimise | Whether higher values are better or lower values are better for the chosen objective. |
| Black-box function | A system where you can choose inputs and observe outputs, but you do not know the true equation underneath. Much of chemistry can be descrribed as black-box |
| Search space | The set of allowed input values BO is allowed to explore. |
| Bounds | The allowed numerical range for a continuous variable. |
| Feature | In the context of BO, an input variable, such as temperature, residence time, or ligand. |
| Target | The output that BO is trying to model or optimise. |
| Iteration | One BO cycle: fit or update the model, score candidates, select a batch of experiments, then collect their results. |
| Candidate point | A possible next experiment considered by the acquisition function. |
| Batch BO | Select several experiments before receiving any of their results. Batch size 1 gives sequential BO. |


## Surrogate-model terms

| Term | Meaning |
| --- | --- |
| Surrogate model | A fast statistical model used in place of the real experiment. We use GPs here.|
| Oracle | In these workbooks, a trained surrogate model wrapped as a callable function. |
| Gaussian Process (GP) | A probabilistic model that predicts both a mean and an uncertainty. |
| Mean prediction | The model's best guess for the target value. |
| Standard deviation | The model's uncertainty about that prediction. |
| Kernel | The GP component that controls how smooth or flexible the fitted response surface is, and how one data point informs another. |
| Matern kernel | A common GP kernel that allows moderately rough response surfaces. |
| RBF kernel | A smoother GP kernel that assumes the response changes very smoothly. |
| Rational Quadratic (RQ) | A kernel that can mix multiple length scales. |
| Periodic kernel | A kernel for repeating patterns. |


## Acquisition-function terms

| Term | Meaning |
| --- | --- |
| Acquisition function | The rule BO uses to decide which candidate experiment to run next. |
| Exploration | Trying uncertain regions to learn more about the response surface. |
| Exploitation | Focusing on regions that already look promising. |
| Expected Improvement (EI) | Scores points by how much they are expected to improve on the current best result. |
| Probability of Improvement (PI) | Scores points by the probability of beating the current best result. |
| Upper Confidence Bound (UCB) | Scores points using predicted mean plus an uncertainty bonus. |
| Greedy acquisition | Uses only the predicted mean, with no explicit uncertainty bonus. |
| Thompson sampling | Samples from the model posterior so the next experiment is chosen stochastically. |
| `EI_XI` | A threshold for EI and PI; larger values demand clearer improvement and therefore increase exploration. |
| `UCB_BETA` | The uncertainty weight in UCB; larger values increase exploration. |


## Data and chemistry terms

| Term | Meaning |
| --- | --- |
| Continuous variable | A numeric variable such as temperature or catalyst loading. |
| Categorical variable | A label-like variable such as ligand, solvent, or base. |
| Encoding | Turning categorical labels into numbers so a model can use them. |
| One-hot encoding | Represent each category by a binary vector with one active position. |
| Ordinal encoding | Represent each category by a simple integer code. |
| Custom encoding | Replace each category with a descriptor vector chosen by the user. |
| Initial design | The first batch of experiments BO sees before it starts using the GP to guide choices. |
| Space-filling design | An initial design chosen to spread points out across the search space. |
| Latin hypercube | A structured way to spread continuous samples across each dimension. Often described as equivalent to a layout where "rooks on a chess board cannot attack each other" |
| Descriptor | A numeric property used to represent a molecule, ligand, or reaction component. |


## Diagnostics and evaluation terms

| Term | Meaning |
| --- | --- |
| Best-so-far curve | The running best objective value seen up to each experiment. |
| Regret | How far the current result is from the best result found so far or from the true optimum, depending on the definition being used. |
| Simple regret | In these notebooks, the gap between the best-so-far value and each point on the curve. |
| Single run | One BO campaign with one random seed. |
| Multi-run summary | Repeating BO across several seeds to see average behaviour and variability. |
| Mean +/- 1 std | A summary band showing the average behaviour plus or minus one standard deviation across runs. |
| Min-max envelope | The full observed range across runs at each experiment index. |
| Seed | The random-number initialisation that controls reproducible randomness. |
| Cross-validation (CV) | Splitting data into folds to estimate model performance more robustly. |


## Notebook terms

| Term | Meaning |
| --- | --- |
| `DATASET_KEY` | The short name used to select a dataset in notebook 2. |
| `ORACLE_KEY` | The short name used to select a prebuilt oracle in workbook 5. |
| `OPTIMIZE_TARGET` | The target column BO should optimise for the current run. |
| `SEARCH_SPACE_OVERRIDE` | A user-defined restriction on the default search space. |
| `N_INITIAL` | Number of experiments in the initial design. |
| `N_ITER` | Number of BO batches after the initial design. |
| `N_CANDIDATES` | Number of candidate points sampled and scored at each BO step. |
| `SAVE_EXPERIMENT_LOGS` | Whether BO writes a CSV log of the evaluated experiments. |
| `scripts/registry.py` | The registry that tells workbook 5 which prebuilt oracle modules are available. |
| `BATCH_SIZE` | Experiments per BO batch. Maximum budget: `N_INITIAL + N_ITER * BATCH_SIZE`. |
